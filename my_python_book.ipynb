## Conditional statements (if, else, and elif)

* **Conditional statements :-** _They are fundamental programming constructs that allow you to control the flow of your program based on conditions that you specify_  
  
![conditional statement flowchart](https://www.guru99.com/images/2013/04/if_then_flowchart.png)

* **__Format :-__**  
    **if**(Expression 1)**:**  
        - Statement  
        - Statement  
        ⇓   
    **elif**(Expression 1)**:**  
        - Statement  
        - Statement  
        ⇓
    **elif**(Expression 1)**:**  
        - Statement  
        - Statement  
        ⇓
    **else:**  
        - Statement  
        - Statement  
        ⇓

````python
#creating a conditional statement
age=int(input("enter your age :"))
if(age<18):
    print("your age is less then 18")
elif(age==18):
    print("your age is 18")
else:
    print("your age is greaer then 18")
````

* **the output of the above program depend's on the input given to veriable age**  
__if():__ execute when input is less then 18

### Adding extra functionality in object from child class

* When you add the __init__() function, the child class will no longer inherit the parent's __init__() function.
* Note: The child's __init__() function overrides the inheritance of the parent's __init__() function.
* To keep the inheritance of the parent's __init__() function, add a call to the parent's __init__() function:
  
```python
class Student(Person):
  def __init__(self, fname, lname,year):
    Person.__init__(fname, lname)
    self.graduationyear = year
    
x = Student("Mike", "Olsen", 2019)
print(x.graduationyear)
```

* As part of polymorphism, a Python child class has methods with the same name as a parent class method.

### Overriding

**overriding** is where you change the behavior of the base class via a function with the same name in a subclass. 
* it is a **runtime potymorphism**

````python
#withour overriding

class parent:
    def p_method(name):
        print("hii ",name,"how are you")
        print("this is parent class")
    
class child(parent):
    def ask():
        print("tell me about your self")
        
s1=child
s1.p_method("rohit")
    
````

````python
# with overriding

class parent:
    def p_method(name):
        print("hii ",name,"how are you")
        print("this is parent class")
    
class child(parent):
    def p_method(name):
        print("hii",name,"this is child class")
    
    def ask():
        print("tell me about your self")
        
s1=child
s1.p_method("rohit")
````

1. As the method **(p_method)** is used in both parent and child class
1. Dispite the inharitance of the parent class in clide class.
1. The method of the parent class would not be called as the child has same method with same perameters
1. This **phenomena** of not calling the parent class if the children class has the same method is known as overriding
1. Overriding can be used to **add the additional properties** to an object

### Overloading

**overloading** is when you define 2 or more methods with the same name but different parameters.  
* it is a **compiletime polymorphism**  
* **python does not support method overloading** by default. But there are different ways to achieve method overloading in Python.
* By Using **Multiple Dispatch** Decorator overloading is possible in python

#### example 1

````python
from multipledispatch import dispatch

class productall:
    # passing one parameter
    @dispatch(int, int)
    def product(self,first, second):
        result = first*second
        print(result)

    # passing two parameters
    @dispatch(int, int, int)
    def product(self,first, second, third):
        result = first * second * third
        print(result)

    # you can also pass data type of any value as per requirement
    @dispatch(float, float, float)
    def product(self,first, second, third):
        result = first * second * third
        print(result)
        
p=productall()
````

````python
p.product(12,45)
````

````python
p.product(65,85,97)
````

````python
p.product(11.1,2.2,4.2)
````
