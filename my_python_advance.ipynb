````python
def make():
    print("please enter the following perameters:")
    name=input("name :")
    age = input("age :")
    joining_date = input("joining-date :")
    Department = input("Department :")
    phone_number = input("phone-number :")
    email = input("email :")
    employ_id = input("employ_id :")
    mk = open("emp_records.txt","a")
    mk.write(name+":"+age+":"+joining_date+":"+Department+":"+phone_number+":"+email+":"+employ_id+"\n")
    mk.close()
````

````python
def search():
    search = input("name of the person you want to search:")
    r = open("emp_records.txt","r")
    lines=r.readlines()
    r.close()
    c=0
    for line in lines:
        if search in line:
            c=c+1
            print(line) 
    if c==0:
        print("data not found")
````

````python
def delete():
    s =input("whose data you want to delete:-")
    d = open("emp_records.txt","r") # read write mode
    lines=d.readlines()
    d.close()
    c=0
    for line in lines:
        if s in line:
            c=c+1
            lines.remove(line)
            d = open("emp_records.txt","w")
            d.writelines(lines)
            print("record remover successfully")
            d.close()
    if c==0:
        print("data not found")
````

````python
def update():
    upd=input("whome you want to update:")
    f = open("emp_records.txt","r")
    lines=f.readlines()
    f.close()
    c=0
    for line in lines:
        if upd in line:
            c=c+1
            lines.remove(line)
            d = open("emp_records.txt","w")
            d.writelines(lines)
            d.close()
            make()
    if c==0:
        print("success fully updated")
            
    
````
